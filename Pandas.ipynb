**DataFrames**

```python
stu={
    "name":["ram","latha","mani"],
    "age":[25,34,55],
    "cgpa":[8,9,10]
}
```

```python
df=pd.DataFrame(stu)
print(df)
```

```python
df=pd.DataFrame(stu,index=['std 1','std 2','std 3'])
print(df)
```

```python
print(df.loc['std 2'])
```

```python
#ADDING Another Column
df['medium']=['english','english','tamil']
print(df)
```

```python
#Adding Another Row
std4=pd.DataFrame({
    "Name":"undefined",
    "age":2,
    "cgpa":8}, index=['std 4']
)
df=pd.concat([df,std4])
print(df)
```

```python
#Adding Multiple Rows
students=pd.DataFrame([
    {"Name":"priya","age":21,"cgpa":7},
    {"Name":"suresh","age":22,"cgpa":7}],index=['std 5','std 6'])
print(pd.concat([df,students]))
```

Data **Cleaning**

```python
df=pd.read_excel('datasets_suresh.xlsx')
```

```python
df=df.drop(columns=['Country'])
print(df)
```

```python
df=df.dropna(subset=['Amount Raised (USD)'])
print(df)
```

```python
df['Amount Raised (USD)']=df['Amount Raised (USD)'].fillna(5)
print(df)
```

```python
df=df.drop(columns=['Industry','Country'])
print(df)
```

```python
#case Sensitive
df['Startup Name']=df['Startup Name'].str.upper()
print(df)
```

```python
#datatype conversion
import numpy as np
df['Number of Employees']=df['Number of Employees'].astype(np.int64)
print(df)
```

```python
#remove duplicates
df=df.drop_duplicates(inplace=True)
```

```python
df.duplicated()
```

```python
df=df.drop_duplicates()
print(df)
```

```python
#missing value
df=df.isnull()
df=df.isnull().sum()
print(df)
```

```python
#use mean,median
df['Amount Raised (USD)'].fillna(df['Amount Raised (USD)'].mean(),inplace=True)
print(df)
```

```python
#remove unwanted spaces
df['Startup Name']=df['Startup Name'].str.strip()
print(df)
```
