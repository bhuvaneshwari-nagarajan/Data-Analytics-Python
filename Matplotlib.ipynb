**Using Arrays in plot**

```python
import numpy as np
x=np.array([2021,2022,2023,2024,2025,2026])
y=np.array([0,10000,20000,30000,50000,80000])
plt.plot(x,y)
plt.show()
```

**Marker and Line Customixation**

```python
import numpy as np
#marker used
x=np.array([2021,2022,2023,2024,2025,2026])
y=np.array([0,10000,20000,30000,50000,80000])
plt.plot(x,y,marker='.',markersize=10,markerfacecolor="red",markeredgecolor="black",markeredgewidth=5,linestyle='dashdot')
plt.show()
```

```python
import numpy as np
x=np.array([2021,2022,2023,2024,2025,2026])
y=np.array([0,10000,20000,30000,50000,80000])
y1=np.array([10000,20000,30000,40000,50000,60000])
plt.plot(x,y,marker='.',markersize=10,markerfacecolor="red",markeredgecolor="black",markeredgewidth=5,linestyle='dashed')
plt.plot(x,y1,marker='.',markersize=10,markerfacecolor="red",markeredgecolor="black",markeredgewidth=5,linestyle='dotted')
plt.show()
```

```python
x=np.array([2021,2022,2023,2024,2025,2026])
y=np.array([0,10000,20000,30000,50000,80000])
y1=np.array([10000,20000,30000,40000,50000,6000])
plotprops=dict(
    marker='.',
    markersize=10,
    markerfacecolor="red",
    markeredgewidth=5,
    linestyle='dashed'
)
plt.plot(x,y,color='blue',**plotprops)
plt.plot(x,y1,color='black',**plotprops)
plt.show()
```

**Title label tricks**

```python
x=np.array([2021,2022,2023,2024,2025,2026])
y=np.array([0,10000,20000,30000,50000,80000])
y1=np.array([10000,20000,30000,40000,50000,6000])
plotprops=dict(
    marker='.',
    markersize=10,
    markerfacecolor="red",
    markeredgewidth=5,
    linestyle='dashed'
)
plt.plot(x,y,color='blue',**plotprops)
plt.plot(x,y1,color='black',**plotprops)

plt.title("Line Graph",fontsize=25,fontweight="bold",color="purple",family="Sans-serif")
plt.xlabel("x axis")
plt.ylabel("y axis")

plt.subplots_adjust(left=0.2,bottom=0.2)

plt.tick_params(axis='both',labelsize=10,labelcolor="red")
plt.grid()
#plt.grid(axis='x',color='black',linestyle='solid',linewidth=1)

plt.show()
```

**Bar Chart**

```python
import matplotlib.pyplot as plt
import numpy as np
x=np.array(['jan','feb','mar','apr','may','jun'])
y=np.array([10000,20000,30000,40000,50000,60000])
plt.bar(x,y,color='black')
#plt.barh(x,y)(horizontal)

plt.title('Bar Chart')
plt.xlabel('month')
plt.ylabel('sales')
plt.show()
```

**Pie Chart**

```python
import matplotlib.pyplot as plt
import numpy as np
x=np.array(['jan','feb','mar','apr','may','jun'])
y=np.array([10000,20000,30000,40000,50000,60000])

plt.pie(y,labels=x,autopct="%1.1f%%",colors=['red','blue','green','purple','yellow','orange'],shadow=True,explode=[0,1,0,0,0,0],startangle=90)

plt.title('Pie Chart')

plt.show()
```

**Scatter Plots**

```python
import matplotlib.pyplot as plt
import numpy as np
x=np.array([12,35,11,5,78,36])
y=np.array([10000,20000,30000,40000,50000,60000])

x1=np.array(['jan','feb','mar','apr','may','jun'])
y1=np.array([10000,20000,30000,40000,50000,60000])

plt.scatter(x,y,color='red',alpha=0.5,edgecolor='black',label='ramya')
plt.scatter(x1,y1,color='blue',alpha=0.5,edgecolor='black',label='suresh')

plt.title("Scatter Plot")

plt.show()
```

**Histogram**

```python
import matplotlib.pyplot as plt
import numpy as np
x=np.array([12,35,11,5,78,36])
plt.hist(x,bins=30,edgecolor='purple',color='purple')
plt.show()
```

**Subplots**

```python
import matplotlib.pyplot as plt
import numpy as np
#figure-canvas
#axes-subplots

#print(plt.subplots(2,2))
figure,axes=plt.subplots(2,2)

x=np.array([2021,2022,2023,2024,2025,2026,2027])
y=np.array([0,1000,2000,30000,60000,80000,100000])

x1=np.array([12,35,11,5,78,36])
y1=np.array([10000,20000,30000,40000,50000,60000])

x2=np.array([20,30,40,50,60,70])
y2=np.array([10000,20000,30000,40000,50000,60000])

x3=np.array([12,35,11,5,78,36])
y3=np.array([10000,20000,30000,40000,50000,60000])

axes[0,0].plot(x,y)
axes[0,1].bar(x1,y1,color='purple')
axes[1,0].plot(x1,y1,color='blue')
axes[1,1].scatter(x2,y2,color='orange')
axes[1,1].scatter(x3,y3,color='black')

plt.show()
```
